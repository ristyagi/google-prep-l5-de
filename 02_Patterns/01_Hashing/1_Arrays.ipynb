##### Lookup by value

```python
def print_array_index_by_value(a):
    for i in range(len(stock_prices)):
        if stock_prices[i] == a:
            return i
    return -1
```

```python
print_array_index_by_value(103)
```

```python
# more pythonic way - using index() method
try:
    index = stock_prices.index(109)
    print(index)
except ValueError:
    print("Price not found")
```

if we lookup by element and return the index of that item, lookup by value will be order of N, O(n), is program will have to travese all the items to get to the right item.

#### Static vs Dynamic array

<div style="width:600px">



</div>

in python list is a dynamic array, in java we have both static and dynamic arrays.
when a static array is created we cannot insert an element on index which is out of initialy created size.

<div style="width:600px">


    
</div>

static array when created, it will be allocated with a fixed memory size.
for dynamic array, it will be allocated with some initial capacity first,

<div style="width:600px">


    
</div>

after 10 elements are inserted and we want to insert new element, it will be allocated 20 new spaces and the it will copy first 10 elements to first 10 spaces of new 20location and then add the 11th item.
when dynamic arrarys grow there is a additional overhead of allocating new memory spaces and copying old data.
once 30elements are inserted and 31st element is to be inserted, it will be allocated a new space of 60 memory spaces and old 30 values will be copied
this is called geometric progression.
