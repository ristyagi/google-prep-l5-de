##### creating a hash function

```python
# Hashmap implementation in python -> ord(char) function gives ascii value of any character in python

def get_hash(key):
    h = 0
    for char in key:
        h = h + ord(char)
    return h % 100
```

```python
get_hash('march 6')
```

##### creating a class for hash table

```python
class HashTable:
    def __init__(self):
        self.MAX = 100 # creating max variable to make a hashmap of 100 elements
        self.arr = [None for i in range(self.MAX)] # creating a placeholder array with 100 elements with each value is None
    
    def get_hash(self, key):
        h = 0
        for char in key:
            h = h + ord(char)
        return h % self.MAX
    
    def add(self, key, val):
        h = self.get_hash(key)
        self.arr[h] = val
        
    def get(self, key):
        h = self.get_hash(key)
        return self.arr[h]
```

```python
t = HashTable() # creaing object of HashTable class
# t.get_hash("march 6")
```

```python
t.add("march 6", 130)
```

##### creating a class for hash table using get and set

```python
class HashTable2:
    def __init__(self):
        self.MAX = 100
        self.arr = [None for i in range(self.MAX)]
        
    def get_hash(self, key):
        h = 0
        for char in key:
            h = h + ord(char)
        return h % self.MAX

    def __setitem__(self, key, val):
        h = self.get_hash(key)
        self.arr[h] = val
    
    def __getitem__(self, key):
        h = self.get_hash(key)
        return self.arr[h]
```

```python
t = HashTable2()
```

```python
class MyContainer:
    def __init__(self, data):
        self.data = data # Internal storage, e.g., a list or dictionary

    def __getitem__(self, key):
        """
        Retrieves an item from the internal data using the provided key/index.
        """
        return self.data[key]
```

```python
# Create an instance of MyContainer
container_instance = MyContainer(['apple', 'banana', 'cherry'])
```

```python
# Access elements using the square bracket notation, which calls __getitem__
print(container_instance[1])
print(container_instance[0:2]) # Also handles slicing
```
